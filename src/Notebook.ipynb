```python
import os, json
rels  = {}
path = '../data/BATS_3.0/'
cats = os.listdir(path)

for cat in cats:
    if not os.path.isdir(path+cat): continue
    for f in os.listdir(path+cat):
        rels[f.split()[0]]=path+cat+'/'+f

RelPairs = {}
for rel in sorted(list(rels.keys())):
    RelPairs[rel]={}
    data = open(rels[rel]).readlines()
    PosSamples = list([[w.strip().lower() for w in d.split('\t')[:2]] for d in data if len(d)!=0])
    
    for p in PosSamples:
        if p[0] not in RelPairs[rel]:   RelPairs[rel][p[0]] = [] 
        RelPairs[rel][p[0]]+=[e.strip() for e in p[1].split('/') if len(e.strip())!=0]
    
    for p in RelPairs[rel]:
        RelPairs[rel][p] = list(set(RelPairs[rel][p]))
    
   
for rel in RelPairs:
    print(rel, len(RelPairs[rel]))
    with open('../data/BATS_3.0.json','w') as fp: 
        json.dump(RelPairs,fp)
```

```python
from tqdm import tqdm
with open('../data/BATS_4.0.json') as fp: 
    RelPairs = json.load(fp)

for rel in tqdm(RelPairs):
    for relt in RelPairs:
        if rel==relt:   continue
        if len(set(RelPairs[rel]) & set(RelPairs[relt]))>0:    print(rel, relt)

```
